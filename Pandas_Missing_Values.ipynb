```python
# handle Missing Values
import pandas as pd
url = "https://raw.githubusercontent.com/rajendra0968jangid/Ds-Arya/main/file2.json"
df = pd.read_json(url)
print(df)
```

```python
#isnull() 
print(df.isnull())
```

```python
#isnull() with sum()
print(df.isnull().sum())
```

```python
#isnull() with sum()--> In pandas cols = 0, row = 1
print(df.isnull().sum(axis=1))
```

```python
print(df.isnull().sum(axis=0))---> cols wise
```

```python
# drop values by row
print(df.dropna())
```

```python
# drop null values
print(df.dropna(axis=1))
```

```python
print(df.dropna(axis=0))
```

```python
#fill
print(df.fillna(0))
```

```python
#fill with 100
print(df.fillna(100))
```

```python
#fill rollno --> 300 | marks-->400
df["roll no"].fillna(300)
print(df)
```

```python
df["roll no"] = df["roll no"].fillna(300)
print(df)
```

```python
#fill marks-->400
df["marks"].fillna(400)
print(df)
```

```python
df["marks"] = df["marks"].fillna(400)
print(df)
```

```python
# Alternative Option
df =pd.read_json(url)
df.fillna({"roll no":200, "marks":500},inplace=True)
print(df)
```

```python
# marks cols --> nan --> mean fillna
df=pd.read_json(url)
mm= df["marks"].mean()
print(mm)
df["marks"] = df["marks"].fillna(mm)
print(df)
```
