```python
# np.where
import numpy as np
# 1d
arr = np.array([10,3,4,80,90,6,7])
print(arr)
 
# np.where
arr_w = np.where(arr>20,"True","False") # (condition)?"True":"False"
print(arr_w)
 
```

```python
arr1 = np.array([10,3,4,80,90,6,7])
print(arr1)

#np.where
result1 = arr1[np.where(arr1>20)]# in index
print(result1)
```

```python
import numpy as np
arr = np.array([11,20,22,40,45,60])
print(arr)

#even index
even_index = np.where(arr % 2 == 0)
print(even_index)

#value
even_index = arr[np.where(arr % 2 == 0)]
print(even_index)


```

```python
#np.where use as replace
arr = np.array([11,20,22,40,45,60])
print(arr)

#np.where
result = np.where(arr>20,100,0)#condition? "True":"False"(Same like ternary operator)
print(result)
```

```python
#np.where for 2d
import numpy as np
arr = np.array([
    [10,20,25],
    [45,50,55],
    [60,20,70]
])

print(arr)

#np.where
result = np.where(arr>40)
print(result) #index of row,index of cols
#1 0 ->45
#1 1 ->50
#1 2 ->55
#2 0 ->60
#2 2 ->70
```

```python
import numpy as np
arr = np.array([
    [10,20,25],
    [45,50,55],
    [60,20,70]
])

print(arr)

#np.where
rows,cols = np.where(arr>40)
print(rows)
print(cols)

#final value
print(arr[rows,cols])
```

```python
#np.concatenate
import numpy as np
arr1 = np.array([10,20,30])
arr2 = np.array([40,50,60])

#np.concatenate
result = np.concatenate((arr1,arr2))
print(result)

result1 = arr1 + arr2
print(result1)
```

```python
#2d array
import numpy as np
arr1 = np.array([
    [10,20],
    [30,40]
])
arr2 = np.array([
    [50,60],
    [7,80]
])

result = np.concatenate((arr1,arr2)) # (0)row -wise
print(result)

result = np.concatenate((arr1,arr2),axis=1) #(1)col - wise
print(result)
```
